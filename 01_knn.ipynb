```python
import numpy as np

X_train = np.array([[0.0, 0.0], [1.0,1.0], [5.0, 5.0], [6.0,6.0]])
y_train = np.array([0, 0, 1, 1])
x = np.array([5.5, 5.5])
```

```python
diffs = X_train - x
print(f"differences: \n{diffs}")
```

```python
dists = np.linalg.norm(diffs, axis=1)
print(dists)
```

```python
sorted_idx = np.argsort(dists)
print(sorted_idx)
```

```python
k = 1
nearest_idx = sorted_idx[:k]
nearest_labels = y_train[nearest_idx]
prediction = np.bincount(nearest_labels).argmax()
print(prediction)
print(nearest_labels)

```

```python
def predict_one(X_train, y_train, x, k):
    diffs = X_train - x
    dists = np.linalg.norm(diffs, axis=1)
    sorted_idx = np.argsort(dists)
    nearest_idx = sorted_idx[:k]
    nearest_labels = y_train[nearest_idx]
    
    return np.bincount(nearest_labels).argmax()
```

```python
print(predict_one(X_train, y_train, np.array([5.5, 5.5]), k=3))
print(predict_one(X_train, y_train, np.array([0.5, 0.5]), k=1))
```

```python
def knn_predict(X_train, y_train, X_test, k):
    results = [predict_one(X_train, y_train, test_point, k) for test_point in X_test]
    return np.array(results)
```

```python
from sklearn.datasets import load_iris
import numpy as np

data = load_iris()
X, y = data.data, data.target

rng = np.random.default_rng(42)
idx = rng.permutation(len(X))
split = int(0.8 * len(X))

train_idx, test_idx = idx[:split], idx[split:]
X_train, y_train = X[train_idx], y[train_idx]
X_test, y_test = X[test_idx], y[test_idx]
```

```python
predictions = knn_predict(X_train, y_train, X_test, 3)
accuracy = (predictions == y_test).mean()
print(accuracy)
print(predictions.shape)
print(y_test.shape)
```
